```python
# ============================================
# CELDA 1: IMPORTACIÓN DE LIBRERÍAS
# ============================================

import os
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from scipy import stats
from scipy.fft import fft2, fftshift
from scipy.ndimage import label, find_objects
import warnings
warnings.filterwarnings('ignore')

# Configurar visualización
%matplotlib inline
plt.rcParams['figure.figsize'] = (12, 8)
sns.set_style('whitegrid')

print("✅ Librerías importadas correctamente")
```

```python
# ============================================
# CELDA 3: FUNCIONES DE PROCESAMIENTO
# ============================================

def load_image(image_path):
    """Carga una imagen en escala de grises"""
    img = cv2.imread(str(image_path), cv2.IMREAD_GRAYSCALE)
    if img is None:
        raise ValueError(f"No se pudo cargar: {image_path}")
    return img

def preprocess_image(image):
    """Preprocesamiento básico de la imagen"""
    # Redimensionar a 512x512 para estandarizar
    resized = cv2.resize(image, (512, 512))
    # Ecualización del histograma para mejorar contraste
    equalized = cv2.equalizeHist(resized)
    # Suavizado Gaussiano para reducir ruido
    smoothed = cv2.GaussianBlur(equalized, (5, 5), 0)
    return smoothed

def segment_lungs(image):
    """Segmenta la región pulmonar usando solo OpenCV"""
    # Binarización con método de Otsu
    _, binary = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    
    # Operaciones morfológicas para limpiar
    kernel = np.ones((5,5), np.uint8)
    cleaned = cv2.morphologyEx(binary, cv2.MORPH_CLOSE, kernel)
    cleaned = cv2.morphologyEx(cleaned, cv2.MORPH_OPEN, kernel)
    
    # Encontrar contornos
    contours, _ = cv2.findContours(cleaned, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    
    if contours:
        # Tomar el contorno más grande (los pulmones)
        largest_contour = max(contours, key=cv2.contourArea)
        mask = np.zeros_like(cleaned)
        cv2.drawContours(mask, [largest_contour], -1, 255, -1)
        roi = cv2.bitwise_and(image, image, mask=mask)
        return roi, mask
    return image, np.zeros_like(image)

def detect_edges(image, method='canny'):
    """Detecta bordes usando diferentes operadores"""
    if method == 'sobel':
        sobelx = cv2.Sobel(image, cv2.CV_64F, 1, 0, ksize=3)
        sobely = cv2.Sobel(image, cv2.CV_64F, 0, 1, ksize=3)
        edges = np.sqrt(sobelx**2 + sobely**2)
        edges = np.uint8(edges / edges.max() * 255)
    elif method == 'canny':
        edges = cv2.Canny(image, 50, 150)
    elif method == 'laplacian':
        edges = cv2.Laplacian(image, cv2.CV_64F)
        edges = np.uint8(np.abs(edges))
    else:
        raise ValueError("Método no soportado")
    return edges

def apply_fft(image):
    """Aplica la Transformada Rápida de Fourier"""
    f_transform = fft2(image)
    f_shift = fftshift(f_transform)
    magnitude_spectrum = np.log(np.abs(f_shift) + 1)
    return magnitude_spectrum

def plot_processing_pipeline(original, processed, roi, edges, fft, save_path=None):
    """Visualiza el pipeline de procesamiento"""
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    
    images = [original, processed, roi, edges, fft]
    titles = ['Original', 'Preprocesada', 'ROI Pulmonar', 'Bordes', 'Espectro FFT']
    
    for idx, (img, title) in enumerate(zip(images, titles)):
        row = idx // 3
        col = idx % 3
        if idx < len(images):
            axes[row, col].imshow(img, cmap='gray')
            axes[row, col].set_title(title, fontsize=14)
            axes[row, col].axis('off')
    
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=300, bbox_inches='tight')
        print(f"✅ Figura guardada: {save_path}")
    plt.show()
    plt.close()

print("✅ Funciones de procesamiento definidas")
```

```python
# ============================================
# CELDA 4: EXTRACCIÓN DE CARACTERÍSTICAS (CORREGIDA - FINAL)
# ============================================

from collections import Counter

class FeatureExtractor:
    """Clase para extraer todas las características de imágenes médicas"""
    
    def __init__(self):
        pass
    
    # ============================================
    # 1. CARACTERÍSTICAS DE INTENSIDAD
    # ============================================
    def extract_intensity_features(self, image):
        """Extrae características de intensidad de píxeles"""
        # Calcular moda de forma segura
        try:
            mode_result = stats.mode(image.ravel())
            if hasattr(mode_result, 'mode'):
                mode_value = mode_result.mode[0] if len(mode_result.mode) > 0 else 0
            else:
                mode_value = mode_result[0][0] if len(mode_result[0]) > 0 else 0
        except:
            mode_value = Counter(image.ravel()).most_common(1)[0][0]
        
        features = {
            'intensity_mean': np.mean(image),
            'intensity_std': np.std(image),
            'intensity_min': np.min(image),
            'intensity_max': np.max(image),
            'intensity_median': np.median(image),
            'intensity_var': np.var(image),
            'intensity_mode': mode_value,
            'percentile_25': np.percentile(image, 25),
            'percentile_75': np.percentile(image, 75)
        }
        return features
    
    # ============================================
    # 2. CARACTERÍSTICAS DE FORMA
    # ============================================
    def extract_shape_features(self, image, mask):
        """Extrae características de forma desde la región pulmonar"""
        contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        
        if not contours:
            return self._empty_shape_features()
        
        contour = max(contours, key=cv2.contourArea)
        area = cv2.contourArea(contour)
        perimeter = cv2.arcLength(contour, True)
        circularity = (4 * np.pi * area) / (perimeter ** 2) if perimeter > 0 else 0
        compacity = (perimeter ** 2) / area if area > 0 else 0
        
        x, y, w, h = cv2.boundingRect(contour)
        aspect_ratio = w / h if h > 0 else 0
        
        hull = cv2.convexHull(contour)
        hull_area = cv2.contourArea(hull)
        solidity = area / hull_area if hull_area > 0 else 0
        extent = area / (w * h) if w * h > 0 else 0
        
        moments = cv2.moments(contour)
        cx = int(moments['m10'] / moments['m00']) if moments['m00'] != 0 else 0
        cy = int(moments['m01'] / moments['m00']) if moments['m00'] != 0 else 0
        
        features = {
            'shape_area': area,
            'shape_perimeter': perimeter,
            'shape_circularity': circularity,
            'shape_compacity': compacity,
            'bbox_x': x,
            'bbox_y': y,
            'bbox_width': w,
            'bbox_height': h,
            'aspect_ratio': aspect_ratio,
            'solidity': solidity,
            'extent': extent,
            'centroid_x': cx,
            'centroid_y': cy
        }
        return features
    
    def _empty_shape_features(self):
        return {
            'shape_area': 0, 'shape_perimeter': 0, 'shape_circularity': 0,
            'shape_compacity': 0, 'bbox_x': 0, 'bbox_y': 0,
            'bbox_width': 0, 'bbox_height': 0, 'aspect_ratio': 0,
            'solidity': 0, 'extent': 0, 'centroid_x': 0, 'centroid_y': 0
        }
    
    # ============================================
    # 3. CARACTERÍSTICAS DE TEXTURA (GLCM) - CORREGIDO
    # ============================================
    def extract_texture_features(self, image):
        """Extrae características de textura usando GLCM"""
        # Reducir a 64 niveles y asegurar rango 0-63
        img_uint8 = np.uint8(image / 255 * 63) if image.dtype != np.uint8 else image
        glcm = self._compute_glcm(img_uint8)
        
        features = {
            'texture_contrast': self._glcm_contrast(glcm),
            'texture_homogeneity': self._glcm_homogeneity(glcm),
            'texture_correlation': self._glcm_correlation(glcm),
            'texture_energy': self._glcm_energy(glcm),
            'texture_asm': self._glcm_asm(glcm),
            'texture_dissimilarity': self._glcm_dissimilarity(glcm),
            'texture_entropy': self._glcm_entropy(glcm)
        }
        return features
    
    def _compute_glcm(self, img):
        """Calcula la matriz de coocurrencia con verificación de límites"""
        # Asegurar que los valores estén en el rango 0-63
        img = np.clip(img, 0, 63)
        glcm = np.zeros((64, 64))
        h, w = img.shape
        for i in range(h-1):
            for j in range(w-1):
                a = int(img[i, j])
                b = int(img[i, j+1])
                # Verificar que estén dentro del rango
                if a < 64 and b < 64:
                    glcm[a, b] += 1
        # Evitar división por cero
        if glcm.sum() > 0:
            glcm = glcm / glcm.sum()
        return glcm
    
    def _glcm_contrast(self, glcm):
        contrast = 0
        for i in range(64):
            for j in range(64):
                contrast += glcm[i, j] * (i - j) ** 2
        return contrast
    
    def _glcm_homogeneity(self, glcm):
        homogeneity = 0
        for i in range(64):
            for j in range(64):
                homogeneity += glcm[i, j] / (1 + abs(i - j))
        return homogeneity
    
    def _glcm_correlation(self, glcm):
        mean_i = np.sum([i * np.sum(glcm[i, :]) for i in range(64)])
        mean_j = np.sum([j * np.sum(glcm[:, j]) for j in range(64)])
        std_i = np.sqrt(np.sum([(i - mean_i)**2 * np.sum(glcm[i, :]) for i in range(64)]))
        std_j = np.sqrt(np.sum([(j - mean_j)**2 * np.sum(glcm[:, j]) for j in range(64)]))
        if std_i * std_j == 0:
            return 0
        correlation = 0
        for i in range(64):
            for j in range(64):
                correlation += glcm[i, j] * (i - mean_i) * (j - mean_j)
        return correlation / (std_i * std_j)
    
    def _glcm_energy(self, glcm):
        return np.sqrt(np.sum(glcm ** 2))
    
    def _glcm_asm(self, glcm):
        return np.sum(glcm ** 2)
    
    def _glcm_dissimilarity(self, glcm):
        dissimilarity = 0
        for i in range(64):
            for j in range(64):
                dissimilarity += glcm[i, j] * abs(i - j)
        return dissimilarity
    
    def _glcm_entropy(self, glcm):
        glcm_normalized = glcm / (glcm.sum() + 1e-10)
        return -np.sum(glcm_normalized * np.log2(glcm_normalized + 1e-10))
    
    # ============================================
    # 4. CARACTERÍSTICAS DE BORDES
    # ============================================
    def extract_edge_features(self, edges):
        edge_pixels = np.sum(edges > 0)
        total_pixels = edges.size
        features = {
            'edge_count': edge_pixels,
            'edge_density': edge_pixels / total_pixels,
            'edge_avg_length': edge_pixels / 10
        }
        return features
    
    # ============================================
    # 5. CARACTERÍSTICAS MORFOLÓGICAS
    # ============================================
    def extract_morphological_features(self, image):
        _, binary = cv2.threshold(image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
        contours, _ = cv2.findContours(binary, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
        
        if not contours:
            return self._empty_morphological_features()
        
        areas = [cv2.contourArea(c) for c in contours]
        
        eccentricities = []
        major_axes = []
        minor_axes = []
        orientations = []
        
        for c in contours:
            if len(c) >= 5:
                ellipse = cv2.fitEllipse(c)
                major_axis = max(ellipse[1])
                minor_axis = min(ellipse[1])
                orientation = ellipse[2]
                major_axes.append(major_axis)
                minor_axes.append(minor_axis)
                orientations.append(orientation)
                if major_axis > 0:
                    eccentricity = np.sqrt(1 - (minor_axis**2 / major_axis**2))
                    eccentricities.append(eccentricity)
                else:
                    eccentricities.append(0)
        
        features = {
            'morph_num_objects': len(contours),
            'morph_num_regions': len(contours),
            'morph_avg_area': np.mean(areas) if areas else 0,
            'morph_max_area': np.max(areas) if areas else 0,
            'morph_min_area': np.min(areas) if areas else 0,
            'morph_eccentricity': np.mean(eccentricities) if eccentricities else 0,
            'morph_major_axis': np.mean(major_axes) if major_axes else 0,
            'morph_minor_axis': np.mean(minor_axes) if minor_axes else 0,
            'morph_orientation': np.mean(orientations) if orientations else 0
        }
        return features
    
    def _empty_morphological_features(self):
        return {
            'morph_num_objects': 0, 'morph_num_regions': 0,
            'morph_avg_area': 0, 'morph_max_area': 0,
            'morph_min_area': 0, 'morph_eccentricity': 0,
            'morph_major_axis': 0, 'morph_minor_axis': 0,
            'morph_orientation': 0
        }
    
    # ============================================
    # 6. CARACTERÍSTICAS FRECUENCIALES (FFT)
    # ============================================
    def extract_frequency_features(self, image):
        f_transform = fft2(image)
        f_shift = fftshift(f_transform)
        magnitude = np.abs(f_shift)
        
        max_magnitude = np.max(magnitude)
        max_idx = np.unravel_index(np.argmax(magnitude), magnitude.shape)
        center = np.array(magnitude.shape) // 2
        freq_dominant = np.sqrt((max_idx[0] - center[0])**2 + (max_idx[1] - center[1])**2)
        energy = np.sum(magnitude**2)
        total_energy = np.sum(magnitude)
        spectral_centroid = np.sum(magnitude * np.arange(magnitude.size).reshape(magnitude.shape)) / total_energy if total_energy > 0 else 0
        
        magnitude_normalized = magnitude / (np.sum(magnitude) + 1e-10)
        spectral_entropy = -np.sum(magnitude_normalized * np.log2(magnitude_normalized + 1e-10))
        
        features = {
            'freq_max_magnitude': max_magnitude,
            'freq_dominant': freq_dominant,
            'freq_energy': energy,
            'freq_spectral_centroid': spectral_centroid,
            'freq_spectral_entropy': spectral_entropy
        }
        return features
    
    # ============================================
    # 7. MOMENTOS DE HU
    # ============================================
    def extract_hu_moments(self, image):
        moments = cv2.moments(image)
        hu_moments = cv2.HuMoments(moments).flatten()
        hu_moments_log = [-np.sign(m) * np.log10(np.abs(m) + 1e-10) for m in hu_moments]
        features = {f'hu_moment_{i+1}': hu_moments_log[i] for i in range(7)}
        return features
    
    # ============================================
    # 8. MÉTODO PRINCIPAL
    # ============================================
    def extract_all_features(self, image, mask=None):
        features = {}
        features.update(self.extract_intensity_features(image))
        
        if mask is not None and np.any(mask):
            features.update(self.extract_shape_features(image, mask))
        
        features.update(self.extract_texture_features(image))
        edges = cv2.Canny(np.uint8(image), 50, 150)
        features.update(self.extract_edge_features(edges))
        features.update(self.extract_morphological_features(image))
        features.update(self.extract_frequency_features(image))
        features.update(self.extract_hu_moments(image))
        
        return features

# Crear instancia
extractor = FeatureExtractor()
print("✅ Extractor de características CORREGIDO inicializado")
```

```python
# ============================================
# CELDA 5: CARGAR IMÁGENES DE EJEMPLO
# ============================================

sample_images = []
sample_labels = []

test_path = data_path / 'test'

for class_name in ['NORMAL', 'PNEUMONIA']:
    class_path = test_path / class_name
    if class_path.exists():
        images = list(class_path.glob('*.jpeg'))[:2]
        for img_path in images:
            img = load_image(img_path)
            sample_images.append(img)
            sample_labels.append(class_name)

print(f"📸 Cargadas {len(sample_images)} imágenes de ejemplo")
```

## RESPUESTAS A LAS PREGUNTAS

### 1. ¿Qué es una radiografía de tórax?
Una radiografia de torax es una imagen diagnostica que utiliza rayos X para visualizar 
las estructuras del torax, incluyendo pulmones, corazon, costillas y diafragma. Es 
una prueba ampliamente utilizada en medicina.

### 2. ¿Qué es la neumonía?
La neumonia es una infeccion aguda del tejido pulmonar que causa inflamacion de los 
alveolos que son los sacos de aire. Puede ser causada por bacterias, virus u hongos. En las
radiografias se manifiestan como opacidades o consolidaciones en las areas afectadas.

### 3. ¿Cuál es el objetivo del dataset?
El dataset Chest X-Ray Images (Pneumonia) tiene como objetivo proporcionar imagenes 
de rayos X de torax etiquetadas para desarrollar sistemas con IA de diagnosticos asistidos 
por computadora que ayuden a identificar automaticamente casos de neumonia de una manera mas rapida y eficiente.

### 4. ¿Qué clases contiene?
El dataset contiene dos clases:
- **NORMAL**: En el normal encontramos radiografias de torax sin evidencia de neumonia.
- **PNEUMONIA**: En la pneumonia encontramos radiografias de torax con evidencia de neumonia.

### 5. ¿Por qué este conjunto de datos requiere un proceso ETL?
Este dataset requiere un proceso ETL porque:
1. Las imagenes estan en diferentes formatos y resoluciones
2. Los datos no estan estructurados para analisis directo
3. Se necesita estandarizar y normalizar las imagenes
4. La extraccion de caracteristicas convierte imagenes en datos numericos
5. Es necesario limpiar imagenes corruptas o de baja calidad
6. Se requiere balanceo de clases para evitar sesgos
7. Los datos deben dividirse adecuadamente para entrenamiento y prueba
